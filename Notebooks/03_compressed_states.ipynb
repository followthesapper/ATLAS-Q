# Compressed Quantum States — Periodic, Product, and MPS

Explore memory scaling and sampling for the three built-in representations:
- **PeriodicState** (O(1) memory for periodic superpositions)
- **ProductState** (O(n) memory)
- **MatrixProductState** (O(n·χ²) memory with canonicalization + sweep sampling)

```python
from quantum_hybrid_system import QuantumClassicalHybrid, PeriodicState, ProductState, MatrixProductState
import math

# PeriodicState
p = PeriodicState(num_qubits=30, period=8)
print("PeriodicState memory:", p.memory_usage(), "bytes")

# ProductState
prod = ProductState(num_qubits=50)
print("ProductState memory:", prod.memory_usage(), "bytes")

# MPS
mps = MatrixProductState(num_qubits=50, bond_dim=16)
print("MPS memory:", mps.memory_usage(), "bytes (χ=16)")

# Sampling examples
print("\nSampling periodic state in computational basis (10 shots):")
print(p.measure(num_shots=10))

print("\nSampling MPS via sweep sampling (20 shots):")
print(mps.sweep_sample(num_shots=20))
```
