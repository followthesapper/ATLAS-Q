# MPS Canonicalization & Sweep Sampling

This notebook shows **left/right canonicalization** and **sweep sampling**,
the correct way to sample from entangled MPS states.

```python
from quantum_hybrid_system import MatrixProductState
import numpy as np

mps = MatrixProductState(num_qubits=20, bond_dim=8)

# Record a few amplitudes before canonicalization
indices = [0, 1, 5, 42, 137]
amps_before = [mps.get_amplitude(i) for i in indices]

# Canonicalize L→R, then R→L
mps.canonicalize_left_to_right()
mps.canonicalize_right_to_left()

amps_after = [mps.get_amplitude(i) for i in indices]

print("Max |Δamp| after canonicalization:",
      max(abs(a-b) for a,b in zip(amps_before, amps_after)))

# Proper sampling
samples = mps.sweep_sample(num_shots=100)
print("Got", len(samples), "samples. Example:", samples[:10])
```
