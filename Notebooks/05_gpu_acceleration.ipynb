# GPU Acceleration — Batched Modular Exponentiation

If CuPy (CUDA) is available, the system uses a custom kernel to
accelerate modular exponentiation and period candidate checks.

```python
from quantum_hybrid_system import QuantumClassicalHybrid

hybrid = QuantumClassicalHybrid(verbose=True, use_gpu=True)

# Compare CPU vs GPU (falls back to CPU if GPU not available)
tests = [(5, 437), (7, 899), (11, 1763)]

for a, N in tests:
    print(f"\n=== a={a}, N={N} ===")
    # CPU
    cpu = hybrid.find_period(a, N)
    print(f"CPU:  r={cpu.period}  method={cpu.method}  time={cpu.time_seconds*1e3:.3f} ms")
    # GPU (or fallback)
    gpu = hybrid.find_period_gpu(a, N)
    print(f"GPU?: r={gpu.period}  method={gpu.method}  time={gpu.time_seconds*1e3:.3f} ms")
```
