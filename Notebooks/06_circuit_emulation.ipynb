# Quantum Circuit Emulation (Tensor Contractions)

Build small circuits with H/X/Y/Z/RX/RY/RZ/CNOT/CZ.
The backend auto-selects separable (Product) or MPS execution.

```python
from quantum_hybrid_system import QuantumClassicalHybrid
import numpy as np
from collections import Counter

hybrid = QuantumClassicalHybrid(verbose=True, use_gpu=False)

# Simple separable circuit
c = hybrid.create_circuit(num_qubits=3)
c.h(0).h(1).rz(2, np.pi/3)
print(c)
state = hybrid.execute_circuit(c, backend="auto")
print("\nMeasure 100 shots:")
counts = Counter(state.measure(num_shots=100))
print(counts)

# Entangling circuit (will push to MPS in a full implementation)
c2 = hybrid.create_circuit(num_qubits=3)
c2.h(0).h(1).cnot(0, 2)  # Note: execute_on_product_state would raise; here we only illustrate gate building
print("\nEntangling circuit description:")
print(c2)
```
