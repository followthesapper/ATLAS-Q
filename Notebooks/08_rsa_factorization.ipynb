# Real-World Use Case — RSA Semiprime Factorization

Demonstrate factoring moderately small RSA-style semiprimes with the hybrid methods.
We verify results and record timings.

```python
from quantum_hybrid_system import QuantumClassicalHybrid
import math, time

def is_semiprime(n):
    c = 0
    x = n
    p = 2
    while p*p <= x and c <= 2:
        while x % p == 0:
            x //= p
            c += 1
        p += 1 if p==2 else 2
    if x > 1:
        c += 1
    return c == 2

# Example semiprimes (choose sizes your machine can handle quickly)
semiprimes = [221, 299, 391, 437, 667, 713, 899, 1147, 1763, 2491]

hybrid = QuantumClassicalHybrid(verbose=True, use_gpu=True)

records = []
for N in semiprimes:
    if not is_semiprime(N):
        print(f"Skipping {N} (not semiprime by quick check)")
        continue
    t0 = time.perf_counter()
    factors = hybrid.factor_number(N)
    dt = time.perf_counter() - t0
    ok = (factors[0]*factors[1] == N) if isinstance(factors, (list, tuple)) and len(factors)==2 else False
    records.append((N, factors, dt))
    print(f"N={N:<5} factors={factors}  ok={ok}  time={dt*1e3:.2f} ms")

print("\nSummary:")
for N, f, dt in records:
    print(f"{N:<5}  {f}  {dt*1e3:.2f} ms")
```
