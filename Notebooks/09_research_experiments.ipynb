# Research Notebook — QFT Peak Sharpness vs Shots

A small experiment to measure how well the analytic QFT sampling recovers period peaks
as a function of the number of shots.

```python
from quantum_hybrid_system import PeriodicState
import numpy as np
import matplotlib.pyplot as plt

n = 10
r = 5
state = PeriodicState(num_qubits=n, period=r)
N = 2**n
peaks = [k * N // r for k in range(r)]

def fraction_near_peaks(shots, tol=2):
    samples = state.measure(num_shots=shots, use_qft=True)
    good = 0
    for s in samples:
        if any(min((s-p) % N, (p-s) % N) <= tol for p in peaks):
            good += 1
    return good / len(samples)

shot_grid = [50, 100, 200, 400, 800, 1600, 3200]
vals = [fraction_near_peaks(m) for m in shot_grid]

plt.figure()
plt.plot(shot_grid, vals, marker='o')
plt.xlabel("shots")
plt.ylabel("fraction within ±2 bins of peaks")
plt.title("QFT peak recovery vs shots")
plt.show()
```
