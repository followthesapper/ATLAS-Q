# Cross-Validation — Quimb / Qiskit (Optional)

If available, we compare small circuits between our backend and external libraries.

```python
import numpy as np
from collections import Counter
from quantum_hybrid_system import QuantumClassicalHybrid

hybrid = QuantumClassicalHybrid(verbose=False, use_gpu=False)
c = hybrid.create_circuit(num_qubits=2)
c.h(0).cnot(0,1)

state = hybrid.execute_circuit(c, backend="auto")
samples = state.measure(num_shots=200)
counts_hybrid = Counter(samples)
print("Hybrid counts (200 shots):", counts_hybrid)

# --- Try Quimb ---
try:
    import quimb as qu
    import quimb.tensor as qtn
    psi0 = qtn.MPS_computational_state('00')
    # build circuit
    circ = qtn.Circuit(2)
    circ.apply_gate('H', 0)
    circ.apply_gate('CNOT', 0, 1)
    vec = circ.psi.to_dense()
    probs = np.abs(vec)**2
    print("Quimb probs:", probs)
except Exception as e:
    print("Quimb not available:", e)

# --- Try Qiskit ---
try:
    from qiskit import QuantumCircuit, Aer, execute
    qc = QuantumCircuit(2,2)
    qc.h(0)
    qc.cx(0,1)
    qc.measure([0,1],[0,1])
    backend = Aer.get_backend('qasm_simulator')
    job = execute(qc, backend, shots=200)
    res = job.result().get_counts()
    print("Qiskit counts:", res)
except Exception as e:
    print("Qiskit not available:", e)
```
