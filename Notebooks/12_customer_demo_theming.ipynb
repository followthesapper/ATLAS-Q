## Flow

> Tip: Run each cell in order; keep outputs on screen for storytelling.

```python
from quantum_hybrid_system import QuantumClassicalHybrid, PeriodicState, ProductState, MatrixProductState

print("👉 1) Instantiate the system")
hybrid = QuantumClassicalHybrid(verbose=False, use_gpu=True)

print("\n👉 2) Compressed states memory snapshot")
p = PeriodicState(24, period=6)
prod = ProductState(24)
mps = MatrixProductState(48, bond_dim=8)
print({"periodic_bytes": p.memory_usage(), "product_bytes": prod.memory_usage(), "mps_bytes": mps.memory_usage()})

print("\n👉 3) Period peaks via analytic QFT sampling")
p2 = PeriodicState(10, period=5)
print("First 20 samples:", p2.measure(num_shots=20, use_qft=True))

print("\n👉 4) Factor a semiprime")
f = hybrid.factor_number(899)
print("899 ->", f)

print("\n👉 5) Optional GPU check (falls back)")
cpu = hybrid.find_period(7, 899)
gpu = hybrid.find_period_gpu(7, 899)
print({"cpu_ms": cpu.time_seconds*1e3, "gpu_ms": gpu.time_seconds*1e3, "r_cpu": cpu.period, "r_gpu": gpu.period})
```
