# Customer Demo (With Logo & One-Click Story)

Optionally drop a logo at `/mnt/data/logo.png` to brand this run.
This notebook:
1. Shows a header with logo (if available)
2. Runs a short sequence (memory snapshot → period peaks → factor a semiprime)
3. Ends with a bar chart "win slide" that you can screenshot

```python
import os, json
import matplotlib.pyplot as plt
from PIL import Image
from quantum_hybrid_system import QuantumClassicalHybrid, PeriodicState, ProductState, MatrixProductState

# Header with logo if present
logo_path = "/mnt/data/logo.png"
if os.path.exists(logo_path):
    try:
        display(Image.open(logo_path))
    except Exception as e:
        print("Logo display failed:", e)
else:
    print("(Optional) Add a logo at /mnt/data/logo.png")

# Run
hybrid = QuantumClassicalHybrid(verbose=False, use_gpu=True)

p = PeriodicState(24, period=6)
prod = ProductState(24)
mps = MatrixProductState(40, bond_dim=8)
mem = {"periodic": p.memory_usage(), "product": prod.memory_usage(), "mps": mps.memory_usage()}
print("Memory snapshot:", mem)

p2 = PeriodicState(10, period=5)
samples = p2.measure(num_shots=200, use_qft=True)
print("QFT samples (first 20):", samples[:20])

factors = hybrid.factor_number(899)
print("899 factors ->", factors)

# "Win slide" chart (matplotlib, default colors)
labels = ["Periodic", "Product", "MPS"]
vals = [mem["periodic"], mem["product"], mem["mps"]]
plt.figure()
plt.bar(labels, vals)
plt.title("Compressed Representations — Relative Memory (bytes)")
plt.ylabel("bytes")
plt.show()
```
