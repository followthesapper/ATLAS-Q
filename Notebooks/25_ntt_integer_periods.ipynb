# NTT Integer Periods — Exact Arithmetic Demo

We generate integer sequences with hidden periods and use an **NTT** (Number-Theoretic Transform)
to recover the dominant period **without floating point**.

```python
import numpy as np
from ntt_features import top_period_ntt

rng = np.random.default_rng(0)
N = 256
r_true = 16
x = (5*np.sin(2*np.pi*np.arange(N)/r_true) + rng.integers(-2,3,size=N)).astype(int)

period, meta, X = top_period_ntt(x)
print({"r_true": r_true, "r_est": int(period), "prime_mod": meta[0], "generator": meta[1]})
```
