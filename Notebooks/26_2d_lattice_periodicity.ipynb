# 2D Lattice Periodicity — Texture Demo

We synthesize checkerboard-like textures with noise and estimate dominant 2D frequencies,
then render a simple 2D "QIH-style" histogram of frequency bins.

```python
import numpy as np
import matplotlib.pyplot as plt
from quantum_hybrid_system.tools_qih.periodic2d_features import topk_2d_freqs, hist2d_from_freqs

rng = np.random.default_rng(0)

def checkerboard(h=128, w=128, px=8, py=10, noise=0.1):
    y, x = np.indices((h, w))
    img = (( (x//px + y//py) % 2 )*2-1).astype(float)
    img += noise * rng.normal(size=img.shape)
    return img

img = checkerboard()
coords, mag = topk_2d_freqs(img, k=4)
H = hist2d_from_freqs(coords, mag.shape, bins=32)
print("Top freq coords:", coords, "hist sum:", float(H.sum()))

plt.figure()
plt.imshow(img, cmap="gray")
plt.title("Synthetic checkerboard")
plt.axis("off")
plt.show()

plt.figure()
plt.plot(H)
plt.title("2D frequency-bin histogram (flattened)")
plt.xlabel("bin"); plt.ylabel("probability")
plt.show()
```
