# Chirp Period Tracking — Windowed QIH

We synthesize a chirp whose period drifts from ~20 to ~80 samples.
We then apply a sliding period estimator (FFT-backed) to track the trend and plot it.

```python
import numpy as np
import matplotlib.pyplot as plt
from chirp_features import synth_chirp, track_periods

x, r_true = synth_chirp(T=6000, r_start=20, r_end=80, noise=0.2, seed=1)
est = track_periods(x, win=256, stride=128, topk=1)

plt.figure()
plt.plot(r_true, label="true r(t)")
# upsample estimated track to length T for visualization
E = np.repeat(est, 128)[:len(r_true)]
plt.plot(E, label="estimated r", alpha=0.8)
plt.title("Chirp period tracking")
plt.xlabel("time"); plt.ylabel("period (samples)"); plt.legend(); plt.show()
```
