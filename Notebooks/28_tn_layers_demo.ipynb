# Tensor-Network Scaffolds — Demo

This notebook shows the shapes and basic behavior of:
- TensorTrainBottleneck (factorized linear)
- TTNBlock (local tree aggregation)
- MERALite (coarse/fine multiscale with residual mix)

```python
import numpy as np
import matplotlib.pyplot as plt

try:
    import torch
    from tn_layers import TensorTrainBottleneck, TTNBlock, MERALite
    B, T, D = 2, 64, 128
    x = torch.randn(B, T, D)
    ttb = TensorTrainBottleneck(d_in=D, d_out=D, m=4, rank=8)
    y = ttb(x)
    print("TTB out:", tuple(y.shape))

    ttn = TTNBlock(d=D, width=3)
    y2 = ttn(y)
    print("TTN out:", tuple(y2.shape))

    mera = MERALite(d=D)
    y3 = mera(y2)
    print("MERA-lite out:", tuple(y3.shape))
except Exception as e:
    print("Torch not available or runtime error:", e)
```
